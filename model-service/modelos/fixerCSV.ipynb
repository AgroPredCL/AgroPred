```python
import pandas as pd

# Load the CSV files as DataFrames
cuartel1 = pd.read_csv('p0s0.csv')
cuartel2 = pd.read_csv('p4s2.csv')
cuartel3 = pd.read_csv('p4s8.csv')
cuartel4 = pd.read_csv('p6s4.csv')

# List of dataframes to simplify processing in a loop
dataframes = {'cuartel1': cuartel1, 'cuartel2': cuartel2, 'cuartel3': cuartel3, 'cuartel4': cuartel4}

# Loop through each DataFrame and convert FechaHora column to the desired format
for name, df in dataframes.items():
    # Convert FechaHora to datetime format, coercing errors to NaT for easier troubleshooting
    df['FechaHora'] = pd.to_datetime(df['FechaHora'], errors='coerce').dt.strftime('%d-%m-%Y %H:%M:%S')
    
    # Drop any rows where FechaHora is NaT
    df.dropna(subset=['FechaHora'], inplace=True)

    # Save each modified DataFrame to a new file to avoid overwriting the original files
    df.to_csv(f'{name}_formatted.csv', index=False)

print("Files saved with updated FechaHora format.")
```
