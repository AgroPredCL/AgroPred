```python
# read csvs

import pandas as pd
import numpy as np
import itertools
```

```python
cuartel1 = pd.read_csv('p0s0.csv')
cuartel2 = pd.read_csv('p6s4.csv')
```

```python
cuartel2.head()
```

```python
# Hacer que todos los datos de todas las columnas de ambos dataframes, tengan como maximo 2 decimales 

#cuartel1 = cuartel1.round(2)   

# Juntar columnas de Fecha y Hora en una sola, para cuartel2, en una nueva columna FechaHora

cuartel2['FechaHora'] = cuartel2['Fecha'] + ' ' + cuartel2['Hora']

# Elimino las columnas Fecha y Hora

cuartel2 = cuartel2.drop(columns=['Fecha', 'Hora'])

# Agregar a cuartel2 las columnas de Nitrogeno, Fosforo, Potasio, pH que estan en cuartel1

cuartel2['Nitrogeno'] = cuartel1['Nitrogeno']
cuartel2['Fosforo'] = cuartel1['Fosforo']
cuartel2['Potasio'] = cuartel1['Potasio']
cuartel2['pH'] = cuartel1['pH']

# Función para rellenar valores vacíos de forma cíclica
def fill_cyclic(df, columns):
    for col in columns:
        values = df[col].dropna().tolist()  # Obtiene los valores no nulos como lista
        if values:  # Asegurarse de que haya valores para ciclar
            cycle_values = itertools.cycle(values)  # Crea un ciclo infinito con los valores
            df[col] = df[col].apply(lambda x: next(cycle_values) if pd.isna(x) else x)  # Rellena valores NaN cíclicamente
    return df

# Definir las columnas a rellenar
columns_to_fill = ['Fosforo', 'Nitrogeno', 'Potasio', 'pH']

# Aplicar la función al DataFrame
cuartel2 = fill_cyclic(cuartel2, columns_to_fill)


# Quiero reordenar las columnas de cuartel2 para que queden en el siguiente orden:
# FechaHora,Nitrogeno,Fosforo,Potasio,Conductividad_Electrica,pH,Humedad,Temperatura y luego las demas columnas

cuartel2 = cuartel2[['FechaHora','Nitrogeno','Fosforo','Potasio','Conductividad_Electrica','pH','Humedad','Temperatura'] + [col for col in cuartel2.columns if col not in ['FechaHora','Nitrogeno','Fosforo','Potasio','Conductividad_Electrica','pH','Humedad','Temperatura']]]
```
