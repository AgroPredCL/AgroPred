```python
import pandas as pd

data = pd.read_csv("p0s0.csv")

data['FechaHora'] = pd.to_datetime(data['FechaHora'], format="%m-%d-%Y %H:%M:%S").dt.strftime("%d-%m-%Y %H:%M:%S")

data.to_csv("p0s0.csv", index=False)
```

```python
data_old = pd.read_csv("p4s2OLD.csv")

# Append " 00:00:00" to each entry in the 'FechaHora' column
data_old['FechaHora'] = data_old['FechaHora'].astype(str) + " 00:00:00"

# Save the modified data back to p0s0OLD.csv
data_old.to_csv("p0s0OLD.csv", index=False)
```
